# tạo 1 mảng numpy có kích thức 150 x 5. giả sử mảng này chứa 150 mẫu dữ liệu 
# của sinh vienen  bao gồm chiều cao cân nặng tuổi , luonmgw và điểm trung bình
# tách mảng dựa trên 4 cột đầu tiên thành 1 biến có tên là x và cột cuối cùng thành y
# tách mảng X thành X train X test trong đó X train chứa 70% dữ liệu và tahcs y thành y train và y test trong đó y train chứa 70% dữ leieuj
# tạo 10 tập dữ liệu không chồng chéo nhau

```python
import numpy as np

# 1. Tạo mảng dữ liệu 150 x 5
# Giả lập dữ liệu sinh viên
np.random.seed(42)  # để kết quả có thể tái lập

height = np.random.randint(150, 190, size=150)     # chiều cao (cm)
weight = np.random.randint(45, 90, size=150)       # cân nặng (kg)
age = np.random.randint(18, 25, size=150)           # tuổi
salary = np.random.randint(5, 30, size=150) * 1e6   # lương (VNĐ)
gpa = np.round(np.random.uniform(2.0, 4.0, size=150), 2)  # điểm TB

data = np.column_stack((height, weight, age, salary, gpa))

print("Kích thước mảng dữ liệu:", data.shape)

# 2. Tách X và y
X = data[:, :4]   # 4 cột đầu
y = data[:, 4]    # cột cuối

print("Shape X:", X.shape)
print("Shape y:", y.shape)

# 3. Chia train / test (70% - 30%)
num_samples = X.shape[0]
train_size = int(0.7 * num_samples)

# Trộn dữ liệu
indices = np.random.permutation(num_samples)

train_indices = indices[:train_size]
test_indices = indices[train_size:]

X_train = X[train_indices]
X_test = X[test_indices]

y_train = y[train_indices]
y_test = y[test_indices]

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

# 4. Tạo 10 tập dữ liệu không chồng chéo
# Mỗi tập gồm 15 mẫu (150 / 10)
num_folds = 10
fold_size = num_samples // num_folds

folds = []

for i in range(num_folds):
    start = i * fold_size
    end = start + fold_size
    fold = data[start:end]
    folds.append(fold)

# Kiểm tra
print("\nSố tập dữ liệu:", len(folds))
print("Kích thước mỗi tập:", folds[0].shape)
```
